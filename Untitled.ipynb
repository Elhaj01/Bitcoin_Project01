```python
from tkinter import *

#main window 

myApp = Tk()

# change App text 
myApp.title("Part manager")

#set dimensions
myApp.geometry("800x500")

#body 

#part

part_text=StringVar()
part_label=Label(myApp,text="Part Name ", font=('bold',14),pady=20)
part_label.grid(row=0,column=0)

part_entry=Entry(myApp,textvariable=part_text)
part_entry.grid(row=0,column=1)

#custumer 

custumer_text=StringVar()
custumer_label=Label(myApp,text=" Custumer ", font=('bold',14),pady=20)
custumer_label.grid(row=0,column=2)

custumer_entry=Entry(myApp,textvariable=custumer_text)
custumer_entry.grid(row=0,column=3)

#Retailer

Retailer_text=StringVar()
Retailer_label=Label(myApp,text=" Retailer ", font=('bold',14),pady=20)
Retailer_label.grid(row=1,column=0)

Retailer_entry=Entry(myApp,textvariable=Retailer_text)
Retailer_entry.grid(row=1,column=1)

#Price

Price_text=StringVar()
Price_label=Label(myApp,text="Price", font=('bold',14),pady=20)
Price_label.grid(row=1,column=2)

Price_entry=Entry(myApp,textvariable=Price_text)
Price_entry.grid(row=1,column=3)
#button 


#Run App infinitely
myApp.mainloop()
```
