```python
#Recherche de solution (difficile)

import time
debu_chrono=time.time()

p= 101

def racine(y):
    e=0
    k=2**10
    for x in range(k):
        for q in range(k):
            if x**2==q*y+p:
                if e==0:
                    print("voici la solution de x :",x)
                    e=1
    if e==0:
        print("no solution trouve")

racine(17)  


time.sleep(2)
fin_chrono=time.time()
total_chrono= fin_chrono - debu_chrono
print("temps d'execution :",total_chrono)
```

```python
# fonction racine_approche

import time
debu_chrono=time.time()

p=101
def racine_approche(y,marge):
    e=0
    k=2**10
    for x in range(k):
        if (x**2-y) % p <= marge:
            if e==0:
                print("voici une solution approche :",x)
                e=1
    if e==0:
        print("pas de solution approche")

racine_approche(17,10)    
time.sleep(2)
fin_chrono=time.time()
total_chrono= fin_chrono - debu_chrono
print("temps d'execution :",total_chrono)
```

```python
#activite2 (outils pour les listes)

#1.fonction additiom des termes de deux listes de meme longeur
def addition(liste1,liste2):
    l_somme=[]
    for i in range(len(liste1)):
        l_somme.append((liste1[i]+liste2[i])%100)
    return l_somme
print(addition([1,2,3,4,5,6],[1,1,1,98,98,98]))
```

```python
#2. fonction qui test si une liste est plus petite que liste_max 
import math
def est_plus_petit(liste,liste_max):
    i=0
    n=min(len(liste_max),len(liste))
    for i in range(n):
        if(liste[i]>liste_max[i]):
            return False
    return True
#test1
print(est_plus_petit([0,0,1,2,3,4],[0,0,5]))
#test2
print(est_plus_petit([0,10,0,1,1],[0,0,5]))

print(est_plus_petit([0,1],[0,2,5]))
```
