```python
from tkinter import *
from tkinter import ttk
from tkinter import messagebox 


root = Tk()
root.title("Blockchain")
root.geometry("1200x700")
root.config(background='#2c3e50')
root.iconbitmap('Bitcoin.ico')
#===== Title frame======

frame1=Frame(root,bg='white',bd=1,relief=SUNKEN)
frame1.place(x=400,y=1,width=400,height=50)


label1=Label(frame1,text="LIVRE DE COMPTE",font=("Bauhaus 93",30),bg="white")
label1.pack()


#====== Enteries frame =====

Enteries_frame=Frame(root,bg='black')
Enteries_frame.place(x=120,y=100,width=1100,height=800)

title= Label(Enteries_frame,text="Transaction:",font=('Calibri',18,'bold'),bg='red',fg='white')
title.place(x=30,y=1)

tranasctvalue=StringVar()
txtTrans= Entry(Enteries_frame,textvariable=tranasctvalue,width=20,font=('Calibri',15))
txtTrans.place(x=30,y=50)

#====== Buttons frame =====

#==== fuctions =====

#activite2 (outils pour les listes)

#1.fonction additiom des termes de deux listes de meme longeur
def addition(liste1,liste2):
    l_somme=[]
    for i in range(len(liste1)):
        l_somme.append((liste1[i]+liste2[i])%100)
    return l_somme
#print(addition([1,2,3,4,5,6],[1,1,1,98,98,98]))


#2. fonction qui test si une liste est plus petite que liste_max 
import math
def est_plus_petit(liste,liste_max):
    i=0
    n=min(len(liste_max),len(liste))
    for i in range(n):
        if(liste[i]>liste_max[i]):
            return False
    return True

#test1
#print(est_plus_petit([0,0,1,2,3,4],[0,0,5]))


#3 fonction qui convertit une chaine de caracteres en une liste...

#Constant globale de longeur des blocs 
N=6

#Constante pour preuve de travail
#Max=[0,0,25]

def phrase_vers_liste(phrase):
    liste = [ord(c)% 100 for c in phrase]
    while len(liste) % N>0:
        liste =[0] +liste 
    return liste 
#test
#print("--Phrase vers liste--")
#ch="Vivi moi !"

#print(ch)
#print(phrase_vers_liste(ch))



#Activité 3 (Fonction de hachage)

#fonction un_tour

N=6
p=[7,11,13,17,19,23]

def un_tour(bloc):
    #addition 
    bloc[1]= (bloc[0]+bloc[1]) %100
    bloc[3]= (bloc[2]+bloc[3]) %100
    bloc[5]= (bloc[4]+bloc[5]) %100
    
    #multiplications
    for i in range(N):
        bloc[i]= (p[i]*bloc[i]+1) % 100
        
    #permutation
    bloc = [bloc[N-1]] + bloc[:N-1]
    return bloc

#test 

#print("---test un tour--- ")
#bloc = [0,1,2,3,4,5]
#print("voici notre bloc :",bloc)
#print("voici la transformation du notre bloc :",un_tour(bloc))




#fonction dix_tours 



def dix_tours(bloc):
    for i in range (10):
        #addition 
        bloc[1]= (bloc[0]+bloc[1]) %100
        bloc[3]= (bloc[2]+bloc[3]) %100
        bloc[5]= (bloc[4]+bloc[5]) %100
    
        #multiplications
        for j in range(N):
            bloc[j]= (p[j]*bloc[j]+1) % 100
        
        #permutation
        bloc = [bloc[N-1]] + bloc[:N-1]
    return bloc  

#test 

#print("--- Voici Deux blocs proches sont transformés en deux blocs très différents--- ")
#bloc1 = [0,1,2,3,4,5]
#print("voici notre bloc :",bloc1)
#print("voici la transformation du notre bloc :",dix_tours(bloc1))



#fonction hachage(liste):



def hachage(liste):
    while len(liste)>N:
        bloc1 = liste[0:N] # Premier bloc
        bloc2 = liste[N:2*N] # Second bloc
        fin_liste = liste[2*N:] # Le reste
    
        bloc1 = dix_tours(bloc1) # Dix tours
    
        nouv_bloc_deb = addition(bloc1,bloc2)
        liste = nouv_bloc_deb + fin_liste
     
    liste = dix_tours(liste) # dix tour pour la dernier bloc
    return liste

# Test
#print("--- Hachage d'une liste ---")
#liste =[0, 1, 2, 3, 4, 5, 12, 3, 24, 72, 47, 77]
#hach = hachage(liste)
#print("voici la liste :", liste)
#print(" l’empreinte de notre liste :",hach)



#Activité 4 (Preuve de travail - Minage)

#1.Vérification (facile)
from time import *


Max=[0,0,5]

def verification_preuve_de_travail(liste,preuve):
    
    concat=liste+preuve
    hach1=hachage(concat)
    
    if est_plus_petit(hach1,Max):
        return True
    else:
        return False

#test
#print("--- Verification de Preuve de travail ---")

#l0=[0,1,2,3,4,5]
#p0=[0,0,2,0,61,2]

#---
#start_time = time()
#print(verification_preuve_de_travail(l0,p0))
#end_time = time()

#duree1 = end_time-start_time

#print("Temps de calcul :",duree1)



#2. Recherche de solution (difficile)

from random import randint

N=6
#Facile
#Max=[0,0,3]

#Moyen
#Max=[0,0,5]

#difficile
#Max=[0,0,0]

def preuve_de_travail(liste):
    hach = [1,1,1,1,1,1]
    while not(est_plus_petit(hach,Max)):
        preuve = [randint(0,99) for i in range(N)]
        liste_test = liste + preuve
        hach = hachage(liste_test)
    #print("voici la solution :", preuve,", et son empreinte :",hach)
    return preuve

#test
#mylist=[0,1,2,3,4,5]
#start_time = time()
#preuve_de_travail(mylist)
#end_time = time()

#duree2 = end_time-start_time

#print("Temps de calcul :",duree2)

#print("----------------------")

#print("la difference entre le temps de calcul d’une simple vérification par rapport au temps de recherche d’une solution :",duree2-duree1)


# Activité 5 - Tes bitcoins

## Question 1 ##
i=1


preuve_init = [0,0,0,0,0,0] # valeur au pif

Livre = [preuve_init]

def ajout_transaction():
    global Livre 
    global i
    
    if i==1:
        Livre = Livre + [tranasctvalue.get()]
        livtext.config(text="bloc numero "+str(i)+": "+str(Livre))
    else:
        Livre = [Livre[-1]]
        Livre = Livre + [tranasctvalue.get()]
        livtext1.config(text="bloc numero "+str(i)+": "+str(Livre))
            
    return Livre
    

# Test
#print("--- Création du livre et ajout d'une transaction ---")
#print(Livre)
#ajout_transaction("ali -150")
#print(Livre)


## Question 2 ##

def minage():
    global Livre
    global i
    if i==1:
        transaction = Livre[-1]
        prec_preuve = Livre[-2]
        liste = prec_preuve + phrase_vers_liste(transaction)
        preuve = preuve_de_travail(liste)
        Livre = Livre + [preuve]
        livtext.config(text="bloc numero "+str(i)+": "+str(Livre))
    else:
        transaction = Livre[-1]
        prec_preuve = Livre[-2]
        liste = prec_preuve + phrase_vers_liste(transaction)
        preuve = preuve_de_travail(liste)
        Livre = Livre + [preuve]
        livtext1.config(text="bloc numero "+str(i)+": "+str(Livre))
        
    return Livre

# Test
#print("--- Minage ---")
#print(Livre)
#minage()
#print(Livre)


## Question 3 ##

def verification_livre():
    prec_preuve = Livre[-3]
    transaction = Livre[-2]
    preuve = Livre[-1]
    hach = hachage(prec_preuve+phrase_vers_liste(transaction)+preuve)
    if est_plus_petit(hach,Max):
        messagebox.showinfo(title='Verification de preuve de travail ',message='la dernière preuve ajoutée au Livre est bien valide')
        
        

    else:
        messagebox.showinfo(title='Verification de preuve de travail ',message='la dernière preuve ajoutée au Livre n est pas valide')

#test
#print("--- Vérification du livre ---")
#print(Livre)
#print(verification_livre())


def Add_new():
    global i
    global livtext1
    livtext1= Label(livre_frame,font=('Gabriola',24,'bold'),fg='white',bg='black')
    livtext1.place(x=5,y=90,width=950,height=30)
    i=i+1

#==== f =====


#=====Button====


btn_frame=Frame(Enteries_frame,bg='black',relief=SOLID)
btn_frame.place(x=30,y=100,width=400,height=40)

btnAdd=Button(btn_frame,
             text="Add",
              width=4,
              height=1,
              font=('Calibri',15),
              fg="white",
              bg="#2980b9",
              bd=0,
              state="normal",
              command=ajout_transaction
             ).place(x=0,y=0)

btnMinage=Button(btn_frame,
             text="Minage",
              width=7,
              height=1,
              font=('Calibri',15),
              fg="white",
              bg="#2980b9",
              bd=0,
              command=minage
             ).place(x=55,y=0)

btnVerifi=Button(btn_frame,
             text="Check",
              width=5,
              height=1,
              font=('Calibri',15),
              fg="white",
              bg="#2980b9",
              bd=0,
              command=verification_livre
             ).place(x=140,y=0)


#====== livre de compte frame =====

livre_frame=Frame(Enteries_frame,bg='white',bd=1,relief=SOLID)
livre_frame.place(x=30,y=170,width=1000,height=300)


title= Label(livre_frame,text="LIVRE DE COMPTE :",font=('Calibri',18,'bold'),bg='red',fg='white')
title.place(x=5,y=1)


Objecttext= Label(livre_frame,text="l'objectif ="+str(Max),font=('Calibri',18,'bold'),bg='yellow')
Objecttext.place(x=210,y=1)

PreufInittext= Label(livre_frame,text="Preuf initiale ="+str(preuve_init),font=('Calibri',18,'bold'),bg='yellow')
PreufInittext.place(x=450,y=1)

livtext= Label(livre_frame,font=('Gabriola',24,'bold'),fg='white',bg='black')
livtext.place(x=5,y=50,width=950,height=30)



btnAdd=Button(Enteries_frame,
              text="Add new bloc",
              width=15,
              height=1,
              font=('Calibri',15),
              fg="white",
              bg="#2980b9",
              bd=0,
              state="normal",
              command=Add_new
             ).place(x=40,y=480)


#=====================================

#=====Message box=====================




root.mainloop()


```
